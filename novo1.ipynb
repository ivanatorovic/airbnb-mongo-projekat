```python
import time
def insert_dataframe_in_batches(df, collection, batch_size=10000):
    total = len(df)
    start = time.time()

    for i in range(0, total, batch_size):
        batch = df.iloc[i:i + batch_size].to_dict("records")
        collection.insert_many(batch)
        print(f"Ubačeno {min(i + batch_size, total)} / {total}")

    end = time.time()
    print(f"Završeno za {round(end - start, 2)} sekundi.\n")



# UBACIVANJE U MONGODB

print("Ubacujem listings...")
insert_dataframe_in_batches(listings, db.listings)

print("Ubacujem listings_detailed...")
insert_dataframe_in_batches(listings_detailed, db.listings_detailed)

print("Ubacujem reviews...")
insert_dataframe_in_batches(reviews, db.reviews)

print("Ubacujem reviews_detailed...")
insert_dataframe_in_batches(reviews_detailed, db.reviews_detailed)

print("Ubacujem neighbourhoods...")
insert_dataframe_in_batches(neighbourhoods, db.neighbourhoods)

print("Ubacujem calendar... ovo može da traje duže.")
insert_dataframe_in_batches(calendar, db.calendar)


# PROVERA BROJA DOKUMENATA

print("Kolekcije u bazi:")
print(db.list_collection_names())

print("\nBroj dokumenata:")
print("calendar:", db.calendar.count_documents({}))
print("listings:", db.listings.count_documents({}))
print("listings_detailed:", db.listings_detailed.count_documents({}))
print("reviews:", db.reviews.count_documents({}))
print("reviews_detailed:", db.reviews_detailed.count_documents({}))
print("neighbourhoods:", db.neighbourhoods.count_documents({}))
```

```python
# U kojim naseljima domaćini imaju najveći broj recenzija? 

import time
import pandas as pd

start = time.time()

rezultat = list(db.reviews.aggregate([

    {
        "$group": {
            "_id": "$listing_id",
            "number_of_reviews": { "$sum": 1 }
        }
    },

    {
        "$lookup": {
            "from": "listings",
            "localField": "_id",
            "foreignField": "id",
            "as": "listing"
        }
    },

    {
        "$unwind": "$listing"
    },

    {
        "$group": {
            "_id": "$listing.neighbourhood",
            "Total reviews": { "$sum": "$number_of_reviews" },
            "Number of listings": { "$sum": 1 }
        }
    },

    {
        "$project": {
            "_id": 0,
            "Neighbourhood": "$_id",
            "Total reviews": 1,
            "Number of listings": 1
        }
    },

    {
        "$sort": {
            "Total reviews": -1
        }
    },

    {
        "$limit": 20
    }

], allowDiskUse=True))

end = time.time()

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

df = pd.DataFrame(rezultat)
df
```

```python
#  Da li verifikovani domaćini imaju veći broj rezervisanih dana od neverifikovanih domaćina?

start = time.time()

rezultat = list(db.calendar.aggregate([

    {
        "$group": {
            "_id": "$listing_id",
            "rezervisani_dani": {
                "$sum": {
                    "$cond": [
                        {"$eq": ["$available", "f"]},
                        1,
                        0
                    ]
                }
            }
        }
    },

    {
        "$lookup": {
            "from": "listings_detailed",
            "localField": "_id",
            "foreignField": "id",
            "as": "listing"
        }
    },

    {
        "$unwind": "$listing"
    },

    {
        "$match": {
            "listing.host_identity_verified": {
                "$in": ["t", "f"]
            }
        }
    },

    {
        "$group": {
            "_id": "$listing.host_identity_verified",
            "Prosečan broj rezervisanih dana": {
                "$avg": "$rezervisani_dani"
            },
            "Ukupan broj rezervisanih dana": {
                "$sum": "$rezervisani_dani"
            },
            "Broj smeštaja": {
                "$sum": 1
            }
        }
    },

    {
        "$project": {
            "_id": 0,
            "Verifikovan identitet": "$_id",
            "Prosečan broj rezervisanih dana": {
                "$round": ["$Prosečan broj rezervisanih dana", 2]
            },
            "Ukupan broj rezervisanih dana": 1,
            "Broj smeštaja": 1
        }
    },

    {
        "$sort": {
            "Verifikovan identitet": -1
        }
    }

], allowDiskUse=True))

end = time.time()

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

df = pd.DataFrame(rezultat)
df
```

```python
# Da li smeštaji sa instant rezervacijom imaju veću prosečnu popunjenost?

start = time.time()

rezultat = list(db.calendar.aggregate([

    {
        "$group": {
            "_id": "$listing_id",
            "ukupno_dana": { "$sum": 1 },
            "rezervisani_dani": {
                "$sum": {
                    "$cond": [
                        { "$eq": ["$available", "f"] },
                        1,
                        0
                    ]
                }
            }
        }
    },

    {
        "$addFields": {
            "popunjenost": {
                "$multiply": [
                    { "$divide": ["$rezervisani_dani", "$ukupno_dana"] },
                    100
                ]
            }
        }
    },

    {
        "$lookup": {
            "from": "listings_detailed",
            "localField": "_id",
            "foreignField": "id",
            "as": "listing"
        }
    },

    { "$unwind": "$listing" },

    {
        "$match": {
            "listing.instant_bookable": {
                "$in": ["t", "f"]
            }
        }
    },

    {
        "$group": {
            "_id": "$listing.instant_bookable",
            "Prosečna popunjenost (%)": {
                "$avg": "$popunjenost"
            },
            "Broj smeštaja": {
                "$sum": 1
            }
        }
    },

    {
        "$project": {
            "_id": 0,
            "Instant rezervacija": "$_id",
            "Prosečna popunjenost (%)": {
                "$round": ["$Prosečna popunjenost (%)", 2]
            },
            "Broj smeštaja": 1
        }
    },

    {
        "$sort": {
            "Instant rezervacija": -1
        }
    }

], allowDiskUse=True))

end = time.time()

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

df = pd.DataFrame(rezultat)
df
```

```python
#Koji sadržaji se najčešće pojavljuju kod smeštaja sa mnogo recenzija?
import time
import pandas as pd

start = time.time()

rezultat = list(db.reviews.aggregate([

    {
        "$group": {
            "_id": "$listing_id",
            "broj_recenzija": { "$sum": 1 }
        }
    },

    {
        "$match": {
            "broj_recenzija": { "$gte": 50 }
        }
    },

    {
        "$lookup": {
            "from": "listings_detailed",
            "localField": "_id",
            "foreignField": "id",
            "as": "listing"
        }
    },

    {
        "$unwind": "$listing"
    },

    {
        "$match": {
            "listing.amenities": { "$ne": None }
        }
    },

    {
        "$project": {
            "broj_recenzija": 1,
            "amenities_lista": {
                "$split": ["$listing.amenities", ","]
            }
        }
    },

    {
        "$unwind": "$amenities_lista"
    },

    {
        "$project": {
            "sadrzaj": {
                "$trim": {
                    "input": "$amenities_lista",
                    "chars": " []\""
                }
            },
            "broj_recenzija": 1
        }
    },

    {
        "$match": {
            "sadrzaj": { "$ne": "" }
        }
    },

    {
        "$group": {
            "_id": "$sadrzaj",
            "broj_smestaja_sa_sadrzajem": { "$sum": 1 },
            "ukupno_recenzija": { "$sum": "$broj_recenzija" }
        }
    },

    {
        "$project": {
            "_id": 0,
            "sadrzaj": "$_id",
            "broj_smestaja_sa_sadrzajem": 1,
            "ukupno_recenzija": 1
        }
    },

    {
        "$sort": {
            "ukupno_recenzija": -1
        }
    },

    {
        "$limit": 20
    }

], allowDiskUse=True))

end = time.time()

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

pd.DataFrame(rezultat)
```

```python
#Koji smeštaji su najteže dostupni za rezervaciju?
import time
import pandas as pd

start = time.time()

rezultat = list(db.calendar.aggregate([

    {
        "$group": {
            "_id": "$listing_id",
            "ukupno_dana": { "$sum": 1 },
            "zauzeti_dani": {
                "$sum": {
                    "$cond": [
                        { "$eq": ["$available", "f"] },
                        1,
                        0
                    ]
                }
            },
            "slobodni_dani": {
                "$sum": {
                    "$cond": [
                        { "$eq": ["$available", "t"] },
                        1,
                        0
                    ]
                }
            }
        }
    },

    {
        "$addFields": {
            "procenat_zauzetosti": {
                "$multiply": [
                    { "$divide": ["$zauzeti_dani", "$ukupno_dana"] },
                    100
                ]
            }
        }
    },

    {
        "$lookup": {
            "from": "listings",
            "localField": "_id",
            "foreignField": "id",
            "as": "listing"
        }
    },

    {
        "$unwind": "$listing"
    },

    {
        "$project": {
            "_id": 0,
            "listing_id": "$_id",
            "naziv": "$listing.name",
            "naselje": "$listing.neighbourhood",
            "tip_sobe": "$listing.room_type",
            "ukupno_dana": 1,
            "zauzeti_dani": 1,
            "slobodni_dani": 1,
            "procenat_zauzetosti": {
                "$round": ["$procenat_zauzetosti", 2]
            }
        }
    },

    {
    "$sort": {
        "procenat_zauzetosti": -1,
        "zauzeti_dani": -1,
         "listing_id": 1
    }
},

    {
        "$limit": 20
    }

], allowDiskUse=True))

end = time.time()

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

pd.DataFrame(rezultat)
```

```python
#Koje karakteristike turisti najčešće ističu u komentarima?
import time
import pandas as pd

start = time.time()

rezultat = list(db.reviews_detailed.aggregate([

    {
        "$match": {
            "comments": { "$ne": None }
        }
    },

    {
        "$project": {
            "komentar": { "$toLower": "$comments" }
        }
    },

    {
        "$group": {
            "_id": None,

            "clean": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "clean" } },
                        1,
                        0
                    ]
                }
            },

            "location": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "location" } },
                        1,
                        0
                    ]
                }
            },

            "metro": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "metro" } },
                        1,
                        0
                    ]
                }
            },

            "quiet": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "quiet" } },
                        1,
                        0
                    ]
                }
            },

            "host": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "host" } },
                        1,
                        0
                    ]
                }
            },

            "comfortable": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "comfortable" } },
                        1,
                        0
                    ]
                }
            },

            "wifi": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "wifi" } },
                        1,
                        0
                    ]
                }
            }
        }
    },

    {
        "$project": {
            "_id": 0,
            "clean": 1,
            "location": 1,
            "metro": 1,
            "quiet": 1,
            "host": 1,
            "comfortable": 1,
            "wifi": 1
        }
    }

], allowDiskUse=True))

end = time.time()

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

pd.DataFrame(rezultat)
```

```python
#Koji smeštaji imaju najbolji turistički skor i najviše stvarnih komentara turista?
import time
import pandas as pd

start = time.time()

rezultat = list(db.reviews_detailed.aggregate([

    {
        "$group": {
            "_id": "$listing_id",
            "broj_komentara": { "$sum": 1 },
            "broj_razlicitih_turista": { "$addToSet": "$reviewer_id" }
        }
    },

    {
        "$project": {
            "broj_komentara": 1,
            "broj_razlicitih_turista": {
                "$size": "$broj_razlicitih_turista"
            }
        }
    },

    {
        "$lookup": {
            "from": "listings_detailed",
            "localField": "_id",
            "foreignField": "id",
            "as": "listing"
        }
    },

    {
        "$unwind": "$listing"
    },

    {
        "$match": {
            "broj_komentara": { "$gte": 50 },
            "listing.review_scores_location": { "$ne": None },
            "listing.review_scores_cleanliness": { "$ne": None },
            "listing.review_scores_value": { "$ne": None }
        }
    },

    {
        "$addFields": {
            "turisticki_skor": {
                "$add": [
                    { "$multiply": ["$listing.review_scores_location", 0.4] },
                    { "$multiply": ["$listing.review_scores_cleanliness", 0.3] },
                    { "$multiply": ["$listing.review_scores_value", 0.3] }
                ]
            }
        }
    },

    {
        "$project": {
            "_id": 0,
            "listing_id": "$_id",
            "naziv": "$listing.name",
            "naselje": "$listing.neighbourhood_cleansed",
            "tip_sobe": "$listing.room_type",
            "ocena_lokacije": "$listing.review_scores_location",
            "ocena_cistoce": "$listing.review_scores_cleanliness",
            "ocena_vrednosti": "$listing.review_scores_value",
            "broj_komentara": 1,
            "broj_razlicitih_turista": 1,
            "turisticki_skor": {
                "$round": ["$turisticki_skor", 2]
            }
        }
    },

    {
        "$sort": {
            "turisticki_skor": -1,
            "broj_komentara": -1
        }
    },

    {
        "$limit": 20
    }

], allowDiskUse=True))

end = time.time()

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

pd.DataFrame(rezultat)
```
