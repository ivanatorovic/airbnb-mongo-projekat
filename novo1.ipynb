```python
import time
def insert_dataframe_in_batches(df, collection, batch_size=10000):
    total = len(df)
    start = time.time()

    for i in range(0, total, batch_size):
        batch = df.iloc[i:i + batch_size].to_dict("records")
        collection.insert_many(batch)
        print(f"Ubačeno {min(i + batch_size, total)} / {total}")

    end = time.time()
    print(f"Završeno za {round(end - start, 2)} sekundi.\n")



# UBACIVANJE U MONGODB

print("Ubacujem listings...")
insert_dataframe_in_batches(listings, db.listings)

print("Ubacujem listings_detailed...")
insert_dataframe_in_batches(listings_detailed, db.listings_detailed)

print("Ubacujem reviews...")
insert_dataframe_in_batches(reviews, db.reviews)

print("Ubacujem reviews_detailed...")
insert_dataframe_in_batches(reviews_detailed, db.reviews_detailed)

print("Ubacujem neighbourhoods...")
insert_dataframe_in_batches(neighbourhoods, db.neighbourhoods)

print("Ubacujem calendar... ovo može da traje duže.")
insert_dataframe_in_batches(calendar, db.calendar)


# PROVERA BROJA DOKUMENATA

print("Kolekcije u bazi:")
print(db.list_collection_names())

print("\nBroj dokumenata:")
print("calendar:", db.calendar.count_documents({}))
print("listings:", db.listings.count_documents({}))
print("listings_detailed:", db.listings_detailed.count_documents({}))
print("reviews:", db.reviews.count_documents({}))
print("reviews_detailed:", db.reviews_detailed.count_documents({}))
print("neighbourhoods:", db.neighbourhoods.count_documents({}))
```

```python
#  Da li verifikovani domaćini imaju veći broj rezervisanih dana od neverifikovanih domaćina?

start = time.time()

rezultat = list(db.calendar.aggregate([

    {
        "$group": {
            "_id": "$listing_id",
            "rezervisani_dani": {
                "$sum": {
                    "$cond": [
                        {"$eq": ["$available", "f"]},
                        1,
                        0
                    ]
                }
            }
        }
    },

    {
        "$lookup": {
            "from": "listings_detailed",
            "localField": "_id",
            "foreignField": "id",
            "as": "listing"
        }
    },

    {
        "$unwind": "$listing"
    },

    {
        "$match": {
            "listing.host_identity_verified": {
                "$in": ["t", "f"]
            }
        }
    },

    {
        "$group": {
            "_id": "$listing.host_identity_verified",
            "Prosečan broj rezervisanih dana": {
                "$avg": "$rezervisani_dani"
            },
            "Ukupan broj rezervisanih dana": {
                "$sum": "$rezervisani_dani"
            },
            "Broj smeštaja": {
                "$sum": 1
            }
        }
    },

    {
        "$project": {
            "_id": 0,
            "Verifikovan identitet": "$_id",
            "Prosečan broj rezervisanih dana": {
                "$round": ["$Prosečan broj rezervisanih dana", 2]
            },
            "Ukupan broj rezervisanih dana": 1,
            "Broj smeštaja": 1
        }
    },

    {
        "$sort": {
            "Verifikovan identitet": -1
        }
    }

], allowDiskUse=True))

end = time.time()

vremena.append({
    "Upit": "3. Verifikovani domaćini",
    "Tip": "Neoptimizovan",
    "Vreme": round(end - start, 2)
})

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

df = pd.DataFrame(rezultat)
df
```

```python
# Da li smeštaji sa instant rezervacijom imaju veću prosečnu popunjenost?

start = time.time()

rezultat = list(db.calendar.aggregate([

    {
        "$group": {
            "_id": "$listing_id",
            "ukupno_dana": { "$sum": 1 },
            "rezervisani_dani": {
                "$sum": {
                    "$cond": [
                        { "$eq": ["$available", "f"] },
                        1,
                        0
                    ]
                }
            }
        }
    },

    {
        "$addFields": {
            "popunjenost": {
                "$multiply": [
                    { "$divide": ["$rezervisani_dani", "$ukupno_dana"] },
                    100
                ]
            }
        }
    },

    {
        "$lookup": {
            "from": "listings_detailed",
            "localField": "_id",
            "foreignField": "id",
            "as": "listing"
        }
    },

    { "$unwind": "$listing" },

    {
        "$match": {
            "listing.instant_bookable": {
                "$in": ["t", "f"]
            }
        }
    },

    {
        "$group": {
            "_id": "$listing.instant_bookable",
            "Prosečna popunjenost (%)": {
                "$avg": "$popunjenost"
            },
            "Broj smeštaja": {
                "$sum": 1
            }
        }
    },

    {
        "$project": {
            "_id": 0,
            "Instant rezervacija": "$_id",
            "Prosečna popunjenost (%)": {
                "$round": ["$Prosečna popunjenost (%)", 2]
            },
            "Broj smeštaja": 1
        }
    },

    {
        "$sort": {
            "Instant rezervacija": -1
        }
    }

], allowDiskUse=True))

end = time.time()

vremena.append({
    "Upit": "4. Instant rezervacija",
    "Tip": "Neoptimizovan",
    "Vreme": round(end - start, 2)
})

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

df = pd.DataFrame(rezultat)
df
```

```python
#Koje karakteristike turisti najčešće ističu u komentarima?
import time
import pandas as pd

start = time.time()

rezultat = list(db.reviews_detailed.aggregate([

    {
        "$match": {
            "comments": { "$ne": None }
        }
    },

    {
        "$project": {
            "komentar": { "$toLower": "$comments" }
        }
    },

    {
        "$group": {
            "_id": None,

            "clean": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "clean" } },
                        1,
                        0
                    ]
                }
            },

            "location": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "location" } },
                        1,
                        0
                    ]
                }
            },

            "metro": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "metro" } },
                        1,
                        0
                    ]
                }
            },

            "quiet": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "quiet" } },
                        1,
                        0
                    ]
                }
            },

            "host": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "host" } },
                        1,
                        0
                    ]
                }
            },

            "comfortable": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "comfortable" } },
                        1,
                        0
                    ]
                }
            },

            "wifi": {
                "$sum": {
                    "$cond": [
                        { "$regexMatch": { "input": "$komentar", "regex": "wifi" } },
                        1,
                        0
                    ]
                }
            }
        }
    },

    {
        "$project": {
            "_id": 0,
            "clean": 1,
            "location": 1,
            "metro": 1,
            "quiet": 1,
            "host": 1,
            "comfortable": 1,
            "wifi": 1
        }
    }

], allowDiskUse=True))

end = time.time()

print("Vreme izvršavanja:", round(end - start, 2), "sekundi")

pd.DataFrame(rezultat)
```
